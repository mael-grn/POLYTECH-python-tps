```python
# Exercice 3

def saisir_nombre():
    while True:
        try:
            return int(input("Entrez un nombre entier : "))
        except ValueError:
            print("Erreur : Veuillez entrer un nombre entier valide.")

def saisir_operande():
    while True:
        try:
            operande = input("Entrez un opérande (+, -, *, /) : ")
            if operande not in ['+', '-', '*', '/']:
                raise ValueError("Opérande invalide.")
            return operande
        except ValueError as e:
            print("Erreur :", e)

nombre1 = saisir_nombre()
operande = saisir_operande()
nombre2 = saisir_nombre()

try:
    if operande == '+':
        resultat = nombre1 + nombre2
    elif operande == '-':
        resultat = nombre1 - nombre2
    elif operande == '*':
        resultat = nombre1 * nombre2
    elif operande == '/':
        resultat = nombre1 / nombre2
    print(f"Le résultat de {nombre1} {operande} {nombre2} est : {resultat}")
except ZeroDivisionError:
    print("Erreur : Division par zéro !")
```

```python
# Exercice 4

def lire_fichier_ligne_par_ligne(nom_fichier):
    try:
        with open(nom_fichier, 'r') as fichier:
            for ligne in fichier:
                print(ligne.strip())
    except FileNotFoundError:
        print("Erreur : Le fichier n'existe pas.")
    except IOError:
        print("Erreur : Problème lors de la lecture du fichier.")

def compte_nombre_mots_fichier(nom_fichier):
    try:
        with open(nom_fichier, 'r') as fichier:
            contenu = fichier.read()
            mots = contenu.split()
            return len(mots)
    except FileNotFoundError:
        print("Erreur : Le fichier n'existe pas.")
        return 0
    except IOError:
        print("Erreur : Problème lors de la lecture du fichier.")
        return 0

def inverser_lignes_fichier(nom_fichier_origine, nom_fichier_destination):
    try:
        with open(nom_fichier_origine, 'r') as fichier:
            lignes = fichier.readlines()
        with open(nom_fichier_destination, 'w') as fichier:
            for ligne in reversed(lignes):
                fichier.write(ligne.rstrip('\n') + '\n')
    except FileNotFoundError:
        print("Erreur : Le fichier n'existe pas.")
    except IOError:
        print("Erreur : Problème lors de la lecture ou de l'écriture du fichier.")

print("Lecture du fichier ligne par ligne :")
lire_fichier_ligne_par_ligne("mon_fichier.txt")
print("nombre de mots : ", compte_nombre_mots_fichier("mon_fichier.txt"))
inverser_lignes_fichier("mon_fichier.txt", "reverse.txt")
print("Fichier inversé créé : reverse.txt")
```

```python
# Exercice 5

def convertir_lignes_fichier_tableau(nom_fichier):
    try:
        with open(nom_fichier, 'r') as fichier:
            tableau = []
            for ligne in fichier:
                tableau.append(ligne.strip())
        return tableau
    except FileNotFoundError:
        print("Erreur : Le fichier n'existe pas.")
        return []
    except IOError:
        print("Erreur : Problème lors de la lecture du fichier.")
        return []


mon_fichier_tableau = convertir_lignes_fichier_tableau("mon_fichier.txt")
reverse_tableau = convertir_lignes_fichier_tableau("reverse.txt")

print("Lignes présentes dans mon_fichier mais pas dans reverse :")
for ligne in mon_fichier_tableau:
    if ligne not in reverse_tableau:
        print(ligne)

print("Lignes présentes dans reverse mais pas dans mon_fichier :")
for ligne in reverse_tableau:
    if ligne not in mon_fichier_tableau:
        print(ligne)

print("Lignes communes aux deux fichiers :")
for ligne in mon_fichier_tableau:
    if ligne in reverse_tableau:
        print(ligne)
```

```python
# Exercice 6

def frequence_mots(nom_fichier, seuil):
    try:
        with open(nom_fichier, 'r') as fichier:
            contenu = fichier.read().lower()
            mots = contenu.split()
            frequence = {}
            mots_distincts = []
            for mot in mots:
                mot = mot.strip('.,!?;"\'()[]{}')  # Nettoyer la ponctuation
                if mot:
                    frequence[mot] = frequence.get(mot, 0) + 1
            for mot in frequence:
                mots_distincts.append(mot)
            for mot, count in frequence.items():
                if count > seuil:
                    with open("mots_frequents.txt", "w") as fichier_sortie:
                        fichier_sortie.write(f"{mot}: {count}\n")

            for mot in frequence:
                print(f"Mot: {mot}, Fréquence: {frequence[mot]}")

            print("Mots distincts dans le fichier :")
            for mot in mots_distincts:
                print(mot)

    except FileNotFoundError:
        print("Erreur : Le fichier n'existe pas.")
    except IOError:
        print("Erreur : Problème lors de la lecture du fichier.")

frequence_mots("mon_fichier.txt", 2)
```
