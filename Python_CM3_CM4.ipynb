# Chapitre 3 - Gestion des Exceptions en Python
Dans ce notebook, nous allons explorer comment gérer les erreurs en Python en utilisant les exceptions.
Nous verrons pourquoi les exceptions sont importantes, comment les utiliser avec `try` et `except`, et nous pratiquerons avec des exercices.

## 1. Qu'est-ce qu'une exception ?
Une **exception** est une erreur qui survient pendant l'exécution d'un programme. Lorsqu'une exception se produit, le programme s'arrête et affiche un message d'erreur. Par exemple :

### Exemple d'une exception commune


````python
with open('mon_fichier.txt', 'r') as fichier2:
    cont = fichier2.readlines()  # Liste de lignes du fichier

with open('reverse.txt', 'w') as fichier:
    for i in range(len(cont)):
        fichier.write(cont[len(cont) - i - 1]+"\n")  # Chaque ligne est déjà une string

with open('reverse.txt', 'r') as fichier:
    reverse = fichier.readlines()  # Lit le fichier inversé
    print(reverse)
````

````python
with open('mon_fichier.txt', 'r') as fichier2:

  cont = fichier2.readlines()

with open('reverse.txt','w') as fichier:

  for i in range(0,len(cont)):

    fichier.write(cont[len(cont)-i-1])


with open('reverse.txt','r') as fichier:

  reverse = fichier.readlines()

  print(reverse)
````

````python
try:
    # Division par zéro, qui provoque une erreur
    result = 10 / 0
except ZeroDivisionError:
    print("Erreur : Division par zéro !")
````


Si on ne gère pas cette erreur avec `try` et `except`, Python affiche une erreur, et le programme s'arrête.

## 2. Syntaxe de base : `try` et `except`
La gestion des exceptions est faite en entourant le code problématique avec un bloc `try` et `except`.
Voici la structure de base :

```python
try:
     # Code pouvant causer une exception
except TypeDeLException:
     # Code exécuté en cas d'exception
```

### Exemple : Gestion d'une erreur de type



````python
try:
    # Essayer de concaténer une chaîne avec un entier (erreur de type)
    result = "Le résultat est : " + 10
except TypeError:
    print("Erreur : Impossible de concaténer une chaîne et un entier !")
````

## 3. Utiliser plusieurs exceptions
On peut gérer différents types d'exceptions en utilisant plusieurs blocs `except`.



````python
try:
    valeur = int("abc")  # Provoque une erreur de conversion
    result = 10 / valeur
except ValueError:
    print("Erreur : Valeur non convertible en entier !")
except ZeroDivisionError:
    print("Erreur : Division par zéro !")
````



## 4. Les blocs `else` et `finally`
- **`else`** : s'exécute si aucune exception n'est levée dans le bloc `try`.
- **`finally`** : s'exécute toujours, qu'il y ait une exception ou non. Il est souvent utilisé pour le nettoyage de ressources.

### Exemple


````python
try:
    num = int(input("Entrez un nombre : "))
    result = 100 / num
except ValueError:
    print("Erreur : Vous devez entrer un nombre.")
except ZeroDivisionError:
    print("Erreur : Division par zéro !")
else:
    print("Le résultat est :", result)
finally:
    print("Fin du bloc try/except.")
````


## 5. Lever ses propres exceptions avec `raise`
Python permet aussi de lever des exceptions manuellement avec `raise`.

````python
def verifier_age(age):
    if age < 0:
        raise ValueError("L'âge ne peut pas être négatif.")
    print("L'âge est valide.")
#testez avec un age negatif
try:
    verifier_age(-5)
except ValueError as e:
    print("Erreur :", e)
````

## 6. Meilleures pratiques avec les exceptions
 - Utilisez des types d'exceptions spécifiques (`ValueError`, `TypeError`, etc.).
 - Évitez les `except` génériques, sauf si c'est nécessaire.
 - Soyez précis dans les messages d'erreur pour aider au débogage.
 - Utilisez `finally` pour libérer des ressources (fichiers, connexions, etc.).

 Si l'utilisateur entre autre chose qu'un nombre entier, affichez un message d'erreur.


````python
import math

def convertir_en_entier():
  try:
    nombre = int(input("Entrez un nombre entier : "))
    return nombre
  except ValueError:
    print("Erreur : Vous devez entrer un nombre entier.")

resultat = convertir_en_entier()
print("Nombre saisi :", resultat)
````

### Exercice 3 : Calcul de la racine carrée
 Écrivez une fonction `racine_carree(n)` qui prend un nombre et retourne sa racine carrée.
 Levez une exception `ValueError` si le nombre est négatif.

````python
import math
def racine_carree(n):
  try:
    if n < 0:
      raise ValueError("Le nombre ne peut pas être négatif.")
    return math.sqrt(n)
  except ValueError as e:
    print("Erreur :", e)
a = 16
resultat = racine_carree(-a)
print("La racine carrée de", a, "est :", resultat)
````

Dans cette partie, nous allons apprendre à manipuler des fichiers en Python.
Nous verrons comment ouvrir, lire, écrire et fermer un fichier, ainsi que des bonnes pratiques pour éviter les erreurs.

#0. Créer et écrire dans un fichier

````python
# Ouvrir un fichier en mode écriture ("w") : il sera créé s'il n'existe pas
with open("mon_fichier.txt", "w") as fichier:
    fichier.write("Bonjour, ceci est un fichier texte créé en Python.\n")
    fichier.write("Deuxième ligne de texte.\n")

print("Fichier créé avec succès !")
````


## 1. Ouvrir et fermer un fichier
En Python, on peut ouvrir un fichier en utilisant la fonction `open()`. Il est important de fermer le fichier après utilisation.
Le mode d'ouverture détermine les opérations possibles : lecture, écriture, etc.

### Syntaxe de base



````python

fichier = open("mon_fichier.txt", 'r')
fichier.close()

# Les modes courants sont :
# - `'r'` : lecture
# - `'w'` : écriture (efface le contenu existant)
# - `'a'` : ajout
# - `'r+'` : lecture et écriture
````

###Exemples

````python
try:
    # Ouverture du fichier en mode lecture
    fichier = open("exemple.txt", "r")
    print("Fichier ouvert avec succès.")
except FileNotFoundError:
    print("Erreur : Le fichier n'existe pas.")
finally:
    fichier.close()  # Fermeture du fichier
    print("Fichier fermé.")
````


## 2. Lire le contenu d'un fichier
Une fois le fichier ouvert en mode lecture, plusieurs méthodes sont disponibles :
 - `read()` : lit tout le contenu du fichier.
 - `readline()` : lit une seule ligne.
 - `readlines()` : lit toutes les lignes et les renvoie sous forme de liste.

### Exemple : `read()`

````python
try:
    with open("exemple.txt", "r") as fichier:
        contenu = fichier.read()
        print("Contenu du fichier :\n", contenu)
except FileNotFoundError:
    print("Erreur : Le fichier n'existe pas.")
````

### Exemple : `readline()` et `readlines()`


````python
try:
    with open("exemple.txt", "r") as fichier:
        # Lire la première ligne
        ligne = fichier.readline()
        print("Première ligne :", ligne)

        # Lire toutes les lignes dans une liste
        toutes_les_lignes = fichier.readlines()
        print("Toutes les lignes :", toutes_les_lignes)
except FileNotFoundError:
    print("Erreur : Le fichier n'existe pas.")
````

## 3. Écrire dans un fichier
Pour écrire dans un fichier, utilisez le mode `'w'`, `'a'` ou `'r+'`.
 - `write()` : pour écrire du texte.
 - `writelines()` : pour écrire plusieurs lignes (une liste de chaînes).

### Exemple : Écriture avec `write()`

````python
try:
    with open("nouveau_fichier.txt", "w") as fichier:
        fichier.write("Ceci est un nouveau fichier.\n")
        fichier.write("Écriture de texte dans le fichier.\n")
        print("Écriture réussie.")
except IOError:
    print("Erreur : Impossible d'écrire dans le fichier.")
````

### Exemple : Ajout avec `a`


````python
try:
    with open("nouveau_fichier.txt", "a") as fichier:
        fichier.write("Ligne ajoutée en mode 'a'.\n")
        print("Ajout réussi.")
except IOError:
    print("Erreur : Impossible d'ajouter au fichier.")
````

## 4. Utilisation du bloc `with`
Le bloc `with` est recommandé car il ferme automatiquement le fichier une fois les opérations terminées, même s'il y a une erreur.

### Exemple : Lecture avec `with`

````python
try:
    with open("nouveau_fichier.txt", "r") as fichier:
        contenu = fichier.read()
        print("Contenu du fichier (avec with) :\n", contenu)
except FileNotFoundError:
    print("Erreur : Le fichier n'existe pas.")
````

## 5. Meilleures pratiques
 - Utilisez toujours le bloc `with` pour ouvrir les fichiers.
 - Vérifiez si le fichier existe avant de lire pour éviter les erreurs.
 - Soyez prudent avec le mode `'w'` qui écrase le contenu existant.
 - Gérez les exceptions pour rendre votre code robuste.
